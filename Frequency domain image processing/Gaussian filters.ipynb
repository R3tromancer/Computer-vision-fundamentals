```python
#Gauss Low


import cv2
import numpy as np
import matplotlib.pyplot as plt

filterVals = [5, 20, 50, 100, 400]

img = cv2.imread('test.png', cv2.IMREAD_GRAYSCALE)

f = np.fft.fft2(img)
fshift = np.fft.fftshift(f)

rows, cols = img.shape
crow, ccol = rows // 2, cols // 2

for filterVal in filterVals:
    mask = np.zeros((rows, cols), np.float32)
    for u in range(rows):
        for v in range(cols):
            D = np.sqrt((u - crow)**2 + (v - ccol)**2)
            mask[u, v] = np.exp(-(D**2) / (2 * (filterVal**2)))

    fshift_filtered = fshift * mask
    f_ishift = np.fft.ifftshift(fshift_filtered)
    img_back = np.fft.ifft2(f_ishift)
    img_back = np.abs(img_back)

    img_back = cv2.normalize(img_back, None, 0, 255, cv2.NORM_MINMAX)
    img_back = img_back.astype(np.uint8)

    cv2.imwrite(f'gaussLow_{filterVal}.png', img_back)
```

```python
#Gauss High


import cv2
import numpy as np
import matplotlib.pyplot as plt

filterVals = [5, 20, 50, 100, 400]

img = cv2.imread('test.png', cv2.IMREAD_GRAYSCALE)

f = np.fft.fft2(img)
fshift = np.fft.fftshift(f)

rows, cols = img.shape
crow, ccol = rows // 2, cols // 2

for filterVal in filterVals:
    mask = np.zeros((rows, cols), np.float32)
    for u in range(rows):
        for v in range(cols):
            D = np.sqrt((u - crow)**2 + (v - ccol)**2)
            mask[u, v] = 1 - np.exp(-(D**2) / (2 * (filterVal**2)))  # ← only this line changed

    fshift_filtered = fshift * mask
    f_ishift = np.fft.ifftshift(fshift_filtered)
    img_back = np.fft.ifft2(f_ishift)
    img_back = np.abs(img_back)

    img_back = cv2.normalize(img_back, None, 0, 255, cv2.NORM_MINMAX)
    img_back = img_back.astype(np.uint8)

    cv2.imwrite(f'gaussHigh_{filterVal}.png', img_back)
```
