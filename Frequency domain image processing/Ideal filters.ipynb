```python
#Ideal Low

import cv2
import numpy as np
import matplotlib.pyplot as plt

filterVals = [5, 20, 50, 100, 400]
img = cv2.imread('test.png', cv2.IMREAD_GRAYSCALE)

fou = np.fft.fft2(img)
foushift = np.fft.fftshift(fou)

rows, cols = img.shape
crow, ccol = rows // 2 , cols // 2

for filterVal in filterVals:
  filter = np.zeros((rows, cols), np.uint8)
  for u in range(rows):
    for v in range(cols):
        D = np.sqrt((u - crow)**2 + (v - ccol)**2)
        if D <= filterVal:
            filter[u, v] = 1
  fshift_filtered = foushift * filter
  f_ishift = np.fft.ifftshift(fshift_filtered)
  img_back = np.fft.ifft2(f_ishift)
  img_back = np.abs(img_back)
  img_back_norm = cv2.normalize(img_back, None, 0, 255, cv2.NORM_MINMAX)
  img_back_uint8 = img_back_norm.astype(np.uint8)
  cv2.imwrite(f'IdealLow_{filterVal}.png', img_back)
```
