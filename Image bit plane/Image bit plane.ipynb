```python
from PIL import Image
import numpy as np

def PictureProcess(image_path):
    img = Image.open(image_path)
    imgAr = np.array(img)

    filter = 1  # 00000001

    for bit in range(8):
        newPixels = imgAr & filter
        newPixels = (newPixels > 0) * 255

        new_img = Image.fromarray(newPixels.astype(np.uint8))
        new_img.save(f"bit_plane_{bit}.png")

        filter <<= 1

PictureProcess("pic.jpg")
```
